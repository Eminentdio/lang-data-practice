```python


# 1. The Corpus
# A short mock corpus demonstrating different semantic uses of 'kọ' (to build/learn/reject)
# and its environment.

yoruba_corpus = """
    Bàbá mi fẹ́ kọ́ ilé tuntun ní ìlú Ìbàdàn. 
    Ọmọdé náà kọ̀ lati jẹ oúnjẹ rẹ̀ lálẹ́ yìí.
    Mo fẹ́ kọ́ ẹ̀kọ́ nípa ìmọ̀ ẹ̀rọ kọ̀m̀pútà.
    Wọ́n ti kọ́ ilé ẹ̀kọ́ gíga sí àdúgbò wa.
    Bí ó bá kọ ìwé náà, yóò rí iṣẹ́.
    Àgbẹ̀ yìí kọ ebè púpọ̀ fún iṣu rẹ̀.
"""

print(f"Corpus loaded:\n {yoruba_corpus}")
```

```python
# 2. Tokenization
# NLTK breaks the string down into individual tokens (words/punctuation)
# so it can analyze their structural positions.

tokens = nltk.word_tokenize(yoruba_corpus.lower())

# We convert the raw list of tokens into an NLTK 'Text' object, 
# which unlocks powerful linguistic methods.

text_object = Text(tokens)

print(f"Tokenized Corpus: {text_object}")
```

```python
# 3. Generating the Concordance (KWIC)
# We ask the machine to find the word 'kọ', and show us the 30 characters 
# to the left and right of it, for every occurrence.

print("---CONCORDANCE FOR 'kọ'---")
text_object.concordance("kọ", width=80, lines=15)

print("---COLLOCATION---")
# Collocations finds pairs of words that occur together unusually often.
text_object.collocations()
```

##### Customed Tokenization and Unicode Preprocessing

```python
import unicodedata

word1 = "kọ"
word2 = "kọ́"

normal_word1 = unicodedata.normalize("NFD", word1)
normal_word2 = unicodedata.normalize("NFD", word2)

print(f"Are there identical in raw Python? {word1 == word2}")
print(f"Base character for word 1: {normal_word1[0]}")
print(f"Base character for word 2: {normal_word2[0]}")
```

```python
from nltk.tokenize import RegexpTokenizer

yoruba_corpus = """
    Bàbá mi fẹ́ kọ́ ilé tuntun ní ìlú Ìbàdàn. 
    Ọmọdé náà kọ̀ lati jẹ oúnjẹ rẹ̀ lálẹ́ yìí.
    Mo fẹ́ kọ́ ẹ̀kọ́ nípa ìmọ̀ ẹ̀rọ kọ̀m̀pútà.
    Wọ́n ti kọ́ ilé ẹ̀kọ́ gíga sí àdúgbò wa.
    Bí ó bá kọ ìwé náà, yóò rí iṣẹ́.
    Àgbẹ̀ yìí kọ ebè púpọ̀ fún iṣu rẹ̀.
"""

standard_tokens = nltk.word_tokenize(yoruba_corpus)
print("Standard NLTK Tokenized Result: \n")
print(standard_tokens)
print("\n------------------\n")
```

```python
custom_pattern = r"\w+(?:'\w+)?|[áàẹọṣíìúùóò\w]+"
custom_tokenizer = RegexpTokenizer(custom_pattern)
custom_token = custom_tokenizer.tokenize(yoruba_corpus)

print("Custom Regexp Tokenizer Output: ")
print(custom_token)
```

```python
clean_token = []

for token in standard_tokens:
    decomposed = unicodedata.normalize("NFD", token)

    stripped_chars = []
    for char in decomposed:
        if char not in ['\u0300', '\u0301']:
            stripped_chars.append(char)
            
    stripped_token = ''.join(stripped_chars)
    final_token = unicodedata.normalize('NFC', stripped_token).lower()
    clean_token.append(final_token)

    print("Pristine Normalize Token")
    print(clean_token)
```

```python
normalised_text_object = Text(clean_token)

print("--- CONCORDANCE FOR 'kọ' --- ")
normalised_text_object.concordance("kọ")
```
