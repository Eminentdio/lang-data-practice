```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
# Sample text to analyze
text = """
     Last week, I mentioned how we overlook our own development to motivate others.How we get lost in the pool of facades that 
     we have no problem with ourselves because we are the ones with the pens and paper.
"""
doc = nlp(text)

print(f"{'Token':<15} | {'Lemma':<12} | {'POS':<6} | {'Morphological Features'}")
print("__" * 75)

for token in doc:
    if not token.is_punct:
        morph_dict = token.morph.to_dict()
        morph_str = ", ".join(f"{k}={v}" for k, v in morph_dict.items()) if morph_dict else "None"
        print(f"{token.text:<15} | {token.lemma_:<12} | {token.pos_:<6} | {morph_str}")
```
