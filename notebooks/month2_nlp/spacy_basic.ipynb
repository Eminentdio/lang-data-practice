#### 1. Load the pre-trained linguistic model

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

#### 2. Process a sample sentence
##### spaCy tokenizes, tags POS, and maps syntactic dependencies in one pass

```python
text = "The boy was happy to have helped the man."
doc = nlp(text)
```

##### 3. Display the structural linguistic features

```python
print(f"{'Token': <15} | {'Coarse POS': <12} | {'Fine POS': <12} | {'Dependency': <12} | {'Stopwords?'}")
print("-" * 64)

for token in doc:
    print(f"{token.text: <15} | {token.pos_: <12} | {token.tag_:<10} | {token.dep_: <12} | {token.is_stop}")
```

##### Visualizing the POS Tagging

```python
from spacy import displacy

options = {
    "compact": True,            # Use square arc
    "distance": 90,             # Horizontal distance between tokens
    "color": "#1f2937",        # Text and arc line color
    "bg": "#f9fafb",           # Background color
    "font": "Source Sans Pro"
}

displacy.render(doc, style='dep', jupyter=True, options=options)
```

##### Yorùbá POS Tagging Using Blank Language Model (BLM)

```python
nlp_yo = spacy.blank('yo')

yor_text = 'Ìyá àgbà bá Àníkẹ́ wi ní àná fún pípẹ́ tí ó pẹ́ wọlé.'

yor_doc = nlp_yo(yor_text)

print(f"{'Token': <15} | {'Coarse POS': <12} | {'Fine POS': <12} | {'Dependency': <12} | {'Stopwords?'}")
print("-" * 64)

for token in yor_doc:
    print(f"{token.text: <15} | {token.pos_: <12} | {token.tag_:<10} | {token.dep_: <12} | {token.is_stop}")
```
